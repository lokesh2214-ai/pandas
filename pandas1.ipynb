***Create the Employee DataFrame***

```python
import pandas as pd

data = {
    'EmpID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Arun', 'Priya', 'Kumar', 'Divya', 'Rahul',
             'Meena', 'Sathish', 'Nila', 'Vijay', 'Kavya'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'IT',
                   'HR', 'Finance', 'IT', 'Sales', 'Sales'],
    'Age': [25, 29, 32, 27, 24, 35, 31, 28, 30, 26],
    'Salary': [35000, 42000, 58000, 39000, 30000,
               62000, 51000, 47000, 44000, 36000],
    'Experience': [2, 5, 8, 3, 1, 10, 7, 4, 6, 2],
    'City': ['Chennai', 'Madurai', 'Chennai', 'Coimbatore',
             'Salem', 'Chennai', 'Madurai', 'Coimbatore',
             'Chennai', 'Salem']
}

df = pd.DataFrame(data)

print(df)
```

***Question 1: Load and Inspect the Data***

```python
print("First 5 rows:")
print(df.head())

print("\nLast 3 rows:")
print(df.tail(3))

print("\nNumber of rows and columns:")
print(df.shape)

print("\nColumn data types:")
print(df.dtypes)
```

***Question 2: Filter Employee Records***

```python
result = df[(df['Department'] == 'IT') & (df['Salary'] > 40000)]
print(result)
```

***Question 3: Sort the Data***

```python
result = df.sort_values('Salary', ascending=False)
print(result[['Name', 'Department', 'Salary']])
```

***Question 4: Calculate Summary Statistics***

```python
print("Mean Salary:", df['Salary'].mean())
print("Maximum Salary:", df['Salary'].max())
print("Minimum Salary:", df['Salary'].min())
print("Median Salary:", df['Salary'].median())
print("Average Age:", df['Age'].mean())
```

***Question 5: Group by Department***

```python
result = df.groupby('Department').agg(
    EmployeeCount=('EmpID', 'count'),
    AverageSalary=('Salary', 'mean'),
    HighestSalary=('Salary', 'max')
)

result = result.sort_values('AverageSalary', ascending=False)
print(result)
```

***Question 6: Create a New Column***

```python
df['AnnualSalary'] = df['Salary'] * 12

print(df[['Name', 'Salary', 'AnnualSalary']])
```

***Question 7: Find Missing and Duplicate Values***

```python


import pandas as pd


df_copy = df.copy()

df_copy.loc[0, 'City'] = None


df_copy = pd.concat(
    [df_copy, df_copy.iloc[[1]]],
    ignore_index=True
)


print("Missing values:")
print(df_copy.isnull().sum())


df_copy['City'] = df_copy['City'].fillna('Unknown')


df_copy = df_copy.drop_duplicates()


print("\nCleaned data:")
print(df_copy[['EmpID', 'Name', 'City']])

print("\nTotal rows:", len(df_copy))
```

***Question 8: Use Conditional Logic***

```python
df['SalaryLevel'] = df['Salary'].apply(
    lambda x: 'High' if x >= 50000
    else 'Medium' if x >= 40000
    else 'Low'
)

print(df[['Name', 'Salary', 'SalaryLevel']])
```

***Question 9: Work with a Second DataFrame***

```python
marks = pd.DataFrame({
    'EmpID': [101, 103, 105, 108, 110],
    'TrainingScore': [85, 90, 78, 88, 92]
})

result = pd.merge(df, marks, on='EmpID', how='left')
print(result[['EmpID', 'Name', 'TrainingScore']])
```

***Question 10: Answer a Business Question***

```python

average_salary = df['Salary'].mean()


department_salary = df.groupby('Department')['Salary'].mean()


highest_department = department_salary.idxmax()

print(
    "Department with highest average salary:",
    highest_department
)

print(
    "Highest average salary:",
    department_salary.max()
)



print("\nEmployees earning above average salary:")

result = df[df['Salary'] > average_salary]

print(result[['Name', 'Department', 'Salary']])
```
